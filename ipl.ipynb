```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_csv("IPL IMB381IPL2013.csv")

# Count number of players from each country
country_counts = df['COUNTRY'].value_counts()

# Plot the bar chart
country_counts.plot(kind='bar', figsize=(10,6), color='mediumseagreen')

# Add labels and title
plt.xlabel("Country")
plt.ylabel("Number of Players")
plt.title("Distribution of IPL Players by Country")
plt.xticks(rotation=45)
plt.tight_layout()

# Show the chart
plt.show()
```

```python
df['TEAM'].value_counts().plot(kind='bar', figsize=(10,6), color='coral')
plt.title("Number of Players per IPL Team")
plt.xlabel("Team")
plt.ylabel("Player Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.groupby('AUCTION YEAR')['SOLD PRICE'].mean().plot(kind='line', marker='o', color='navy')
plt.title("Average Sold Price Over Auction Years")
plt.xlabel("Auction Year")
plt.ylabel("Average Sold Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df['AGE'].dropna().plot(kind='hist', bins=10, color='slateblue', edgecolor='black')
plt.title("Age Distribution of Players")
plt.xlabel("Age")
plt.ylabel("Number of Players")
plt.tight_layout()
plt.show()
```

```python
df['PLAYING ROLE'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(8,8), startangle=140)
plt.title("Distribution of Playing Roles")
plt.ylabel("")  # Hide y-label
plt.show()
```

```python
df.plot(kind='scatter', x='BASE PRICE', y='SOLD PRICE', color='darkgreen', alpha=0.6)
plt.title("Base Price vs Sold Price")
plt.xlabel("Base Price")
plt.ylabel("Sold Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df.boxplot(column='SR-B', by='PLAYING ROLE', figsize=(10,6), grid=False)
plt.title("Batting Strike Rate by Playing Role")
plt.suptitle("")  # Remove default title
plt.xlabel("Playing Role")
plt.ylabel("Strike Rate")
plt.tight_layout()
plt.show()
```
